```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt 
```

```python
# Load trained model
model = YOLO("Yolov11best.pt")
```

```python
# Open webcam
cap = cv2.VideoCapture(0)
cap.set(cv2.CAP_PROP_FRAME_HEIGHT,1920)
cap.set(cv2.CAP_PROP_FRAME_WIDTH,1080)
```

```python
while cap.isOpened():

    ret, frame = cap.read()

    if not ret:
        break

    # YOLO Prediction
    results = model.predict(
        source=frame,
        conf=0.5,
        verbose=False
    )

    # Draw detections
    annotated_frame = results[0].plot()

    # Display
    cv2.imshow("Rock Paper Scissors - Real Time", annotated_frame)

    # Press Q to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
